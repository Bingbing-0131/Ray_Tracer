```python
import os
with open('./sphere.obj','r') as f:
    with open('./bunny_200_new.obj','w') as f_new:
            # 读取文件内容
        lines = f.readlines()
        for line in lines:
            if line.startswith('v '):
                v = line.split(' ')
                for i in range(1,len(v)):
                    v[i] = "{:.6f}".format(float(v[i])*10+50)
                    
                line = ' '.join(v)
                f_new.write(line+'\n')
            if line.startswith('f'):
                f = line.split(' ')
                tmp = f[-1]
                f[-1] = f[-2]
                f[-2] = tmp
                line = ' '.join(f).replace('\n','')
                f_new.write(line+'\n')
```

```python
with open('wine.txt', 'r') as f:
    with open('wine_new.txt','w') as g:
        lines = f.readlines()
        for line in lines:
            line = line.replace('\n','').replace('[','').replace(']','').strip().split(' ')
            for i in range (3):
                line[i] = float(line[i])*20+75
            line = '[ {:.6f} {:.6f} {:.6f} ]'.format(line[0],line[1],line[2])
            g.write(line+'\n')
```
